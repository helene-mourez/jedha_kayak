```python
df = df.drop(columns=['sys.type', 'sys.id', "sys.country", "coord.lon", "coord.lat", "cod", "base", "name", "id", "timezone", "place_id"])
```

```python
obs_generales = df.groupby(["weather.0.id", "weather.0.main", "weather.0.description"]).agg({"dt": "count"}).reset_index().rename(columns={"dt": "count"})
```

```python
obs_generales
```

```python
villes_météo_top = []

for i, city in enumerate(df["city"]) : 
  if df["weather.0.id"][i] == 800 : 
    villes_météo_top.append(city)
```

```python
villes_météo_top
```

```python
fig = px.bar(obs_generales, x="weather.0.main", y="count", color="weather.0.description", title="Répartition des conditions météorologiques")
```

```python
fig
```

```python
fig = go.Figure(data=[go.Scattergeo(
    lon = df['lon'],
    lat = df['lat'],
    text = df['city'],
    mode = 'markers',
    marker = dict(
        size = 8,
        color = df['weather.0.id'],
        colorscale = 'Viridis',
        colorbar_title = "Weather ID"
    )
    )])
```

```python
fig = go.Figure()
```

```python
fig.add_trace(
    go.Scattermap(
        lat = df["lat"],
        lon = df["lon"],
        mode = "markers",
        marker = dict(
            size = 14,
            color = obs_generales['count'],
            colorscale = "RdYlGn",
            showscale = False,
            opacity = 1,
            colorbar=dict(title="Nombre d'observations", len=0.5, y=0.5)
        ),
        customdata=df[
            ["city", "main.temp", "wind.speed", "weather.0.description", "weather.0.id"]
        ]
    )
)

fig.update_layout(
    map_style ='open-street-map',
    map_center=dict(lat=46.5, lon=2.5),
    map_zoom=4.5,
    height=650,
    margin=dict(l=0, r=0, t=30, b=0)
)

fig.show()
```
