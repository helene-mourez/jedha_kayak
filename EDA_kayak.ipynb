Définition des critères pour générer le top 5 des villes en fonction de la météo :
    - Pondération des critères : 
        - Température ressentie : 0.2
        - Vitesse du vent : 0.15
        - Rafales de vent : 0.2
        - Visibilité : 0.1
        - Couverture nuageuse : 0.35
    - Précipitations : éliminatoire (si pluie > 0, la ville est éliminée)

    - critères de notation :
        - Température ressentie : 
            - 20°C à 25°C : 1 point
            - 10°C à 19°C : 0.5 point
            - 5°C à 10°C : 0 point
            - <5°C : -1 point
        - Vitesse du vent : 
            - <20 km/h : 1 point
            - 20-40 km/h : 0.5 point
            - >40 km/h : 0 point
        - Rafales de vent : 
            - <20 km/h : 1 point
            - 20-40 km/h : 0.5 point
            - >40 km/h : 0 point
        - Visibilité : 
            - >10 km : 1 point
            - 2-10 km : 0.5 point
            - <2 km : 0 point
        - Couverture_nuageuse (weather_id) : 
            - 800 : 1 point
            - 801-802 : 0.5 point
            - 803-804 : 0 point

```python
df_cws = df[["city", "main.feels_like", "wind.speed", "wind.gust", "rain.1h", "weather.0.id"]]
```

```python
df_cws = df_cws.drop_duplicates(subset="city")
```

```python
cities_scores = []

for _,row in df_cws.iterrows():
    city_name = row["city"]
    feeling_temp = row["main.feels_like"]
    wind_speed = row["wind.speed"] * 3.6  # Conversion de m/s en km/h
    wind_gust = row["wind.gust"] * 3.6  # Conversion de m/s en km/h
    rain_1h = row["rain.1h"]
    weather_id = row["weather.0.id"]

    # Calcul du score de température pour chaque ville
    temp_score = 0
    if feeling_temp >= 20 and feeling_temp <= 25:
        temp_score += 1
    elif feeling_temp >= 25 and feeling_temp <= 30:
        temp_score += 0.3
    elif feeling_temp >= 15 and feeling_temp < 20:
        temp_score += 0.5
    elif feeling_temp >= 10 and feeling_temp < 15:
        temp_score += 0.2
    elif feeling_temp >= 5 and feeling_temp < 10:
        temp_score += 0
    elif feeling_temp < 5 or feeling_temp > 30:
        temp_score -= 1

    wind_speed_score = 0
    if wind_speed <= 10:
        wind_speed_score += 1
    elif wind_speed > 10 and wind_speed <= 20:
        wind_speed_score += 0.5
    elif wind_speed >20 and wind_speed <= 30:
        wind_speed_score += 0.3
    elif wind_speed > 30 and wind_speed <= 40:
        wind_speed_score += 0
    elif wind_speed > 40:
        wind_speed_score -= 1

    wind_gust_score = 0
    if wind_gust == 0:
        wind_gust_score += 1
    elif wind_gust > 0 and wind_gust <= 20:
        wind_gust_score += 0.5
    elif wind_gust > 20 and wind_gust <= 30:
        wind_gust_score += 0.3
    elif wind_gust > 30 and wind_gust <= 40:
        wind_gust_score += 0
    elif wind_gust > 40:
        wind_gust_score -= 1

    rain_1h_score = 0
    if rain_1h != 0:
        rain_1h_score -= 10

    cloud_score = 0
    if weather_id == 800:
        cloud_score += 1
    elif weather_id == 801 :
        cloud_score += 0.7
    elif weather_id == 802:
        cloud_score += 0.5
    elif weather_id == 803 :
        cloud_score += 0.3
    elif weather_id == 804:
        cloud_score += 0

    # Calcul du score total pondéré pour chaque ville
    total_score = (
        0.2 * temp_score
        + 0.15 * wind_speed_score
        + 0.2 * wind_gust_score
        + 0.1 * rain_1h_score
        + 0.35 * cloud_score
        )

    cities_scores.append({"city": city_name, "temp_score": temp_score, "wind_speed_score": wind_speed_score, "wind_gust_score": wind_gust_score, "rain_1h_score": rain_1h_score, "cloud_score": cloud_score, "total_score": total_score})
```

```python
cities_scores_df = pd.DataFrame(cities_scores, columns=["city", "temp_score", "wind_speed_score", "wind_gust_score", "rain_1h_score", "cloud_score", "total_score"])
```

```python
cities_scores_df
```

```python
top_5_cities = (
    cities_scores_df
    .sort_values("total_score", ascending=False)
    .head(5)
)

top_5_cities
```

```python
top_5_cities = pd.merge(top_5_cities, df[["city", "lat", "lon"]], on="city", how="left")
```

```python
fig = go.Figure(
    go.Scattermap(
        lat=top_5_cities["lat"],
        lon=top_5_cities["lon"],
        text=top_5_cities["city"],
        mode="markers+text",
        textposition="top center",
        marker=dict(
            size=14,
            color=top_5_cities["total_score"],
            colorscale="Viridis",
            showscale=True,
            colorbar=dict(
                title="Score météo"
            ),
        ),
        hovertemplate=(
            "<b>%{text}</b><br>"
            "Score météo : %{marker.color:.2f}<br>"
            "<extra></extra>"
        ),
    )
)

fig.update_layout(
    map=dict(
        style="open-street-map",
        center=dict(
            lat=46.5,
            lon=2.5,
        ),
        zoom=4.5,
    ),
    height=650,
    margin=dict(l=30, r=30, t=70, b=30),
    title=dict(
        text="Top 5 des villes avec les meilleures conditions météorologiques",
        x=0.5,
        xanchor="center",
    ),
)

fig.show()
```

Le top 5 des villles est défini en fontction de la météo. Les critères de notation sont pondérés pour obtenir un score total pour chaque ville. Les villes avec des précipitations sont éliminées du classement. Le top 5 des villes est ensuite affiché sur une carte avec leurs coordonnées géographiques (latitude et longitude).

!!! Un nouvel affinage des critères de notation et de pondération peut être effectué pour obtenir un classement plus précis et éviter les scores égaux risquant de fausser le classement du top 5. !!!
