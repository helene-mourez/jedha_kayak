```python
import requests
import pandas as pd
from unidecode import unidecode
import time
from get_cities import get_city_info
```

```python
#df = get_city_info("France")
df = pd.read_csv("data/cities.csv")
API_KEY = "95f5ed4b5991c2bd754589456f14cc16"

url = "https://api.openweathermap.org/data/2.5/weather"

rows = []
for _, row in df.iterrows():
    params = {
        "lat": row["lat"],
        "lon": row["lon"],
        "units": "metric",
        "appid": "95f5ed4b5991c2bd754589456f14cc16",
        "lang": "fr"
    }
    response = requests.get(url, params=params)
    # print(row, response.status_code) # check if the call is successful

    payload = response.json()
    # payload["name_city"] = row["name"]

    rows.append(payload)

    time.sleep(1)

df_rows = pd.DataFrame(rows)
print(df_rows)
```

```python
# Applatissement général du JSON pour obtenir un DataFrame plus lisible
df_weather = pd.json_normalize(rows)
```

```python
weather = pd.json_normalize(df_rows["weather"])
final_weather = pd.json_normalize(weather.to_dict(orient="records"))
```

```python
final_weather = final_weather.add_prefix("weather.")
```

```python
display(final_weather.head())
```

```python
df_weather = pd.concat(
    [
        df_weather.drop(columns="weather"),
        final_weather
    ],
    axis=1
)
display(df_weather)
```

```python
df_weather.to_csv("data/cities_weather.csv", index=False) # avoid a second index creation
```
