```python
# Display null values
missing_values = df.isna().sum()
missing_values = missing_values[missing_values > 0] # mask on columns with missing values
print(missing_values)
```

```python
missing_values = df.isnull().mean().sort_values(ascending=False) * 100
missing_values = missing_values[missing_values > 0] # mask on columns with missing values

missing_values_percent = round(len(missing_values) /195 * 100, 2)

print(f"Missing values percentage : {missing_values_percent}\n\nMissing values percentage per column : \n{missing_values}")
```

Let's do a heatmap to visualize missing values distribution.

```python
# display null values distribution
# the figure is more readable yet less representative
# from matplotlib import pyplot as plt
# import seaborn as sns

plt.figure(figsize=(50, 15))
sns.heatmap(df.isnull(), cbar=False)
plt.savefig('nulls.png', dpi=150, bbox_inches='tight')
```
