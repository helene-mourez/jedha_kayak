```python
import os
import logging
from pathlib import Path
from scrapy.utils.reactor import install_reactor
install_reactor("twisted.internet.asyncioreactor.AsyncioSelectorReactor")
import scrapy
from scrapy.crawler import CrawlerProcess
```

```python
class BookingSpider(scrapy.Spider):

    # Name of your spider
    name = "booking"

    # Url to start your spider from 
    start_urls = [
        "https://www.booking.com/index.fr.html?label=gog235jc-10CAEoggI46AdIDVgDaE2IAQGYATO4AQfIAQ3YAQPoAQH4AQGIAgGoAgG4Ar-loNQGwAIB0gIkYjYzZjhkNmMtNzY3MS00ZDZlLWFlYmUtNjU4MWFhMTlkNTE12AIB4AIB&aid=397594&sid=08121f1474c0cb46b3ca9b8928a700ef",
        "https://www.booking.com/searchresults.fr.html?label=msn-8aGmFKcvjur7ORXD1T41Jg-80676853483926%3Atikwd-80677028168879%3Aloc-66%3Anes%3Amte%3Alp186792%3Adec%3Acid518941944%3Aagid1290827950458860%3Aclkida60dbefb136f17da04a9d47302b61cd6&aid=2369666&ss=Valensole%2C+Provence-Alpes-C%C3%B4te+d%27Azur+(Plateau+de+Valensole)%2C+France&efdco=1&lang=fr&src=index&dest_id=-1474456&dest_type=city&ac_position=0&ac_click_type=b&ac_langcode=fr&ac_suggestion_list_length=5&search_selected=true&search_pageview_id=43f73e73bcac020e&checkin=2026-08-24&checkout=2026-08-30&group_adults=1&no_rooms=1&group_children=0&nflt=ht_id%3D204"
    ]

    # Callback function that will be called when starting your spider
    # It will get text, author and tags of all the <div> with class="quote"
    def parse(self, response):
        n = 10
        for i in range(n):
            i = i + 1
            yield {
                'text': response.xpath('/html/body/div/div[2]/div[1]/div[{}]/span[1]/text()'.format(i)).get(),
                'author': response.xpath('/html/body/div/div[2]/div[1]/div[{}]/span[2]/small/text()'.format(i)).get(),
                'tags': response.xpath('/html/body/div/div[2]/div[1]/div[{}]/div/a/text()'.format(i)).getall(),
            }
```

```python
from pathlib import Path

filename = "booking_scrap.json"
output_dir = Path("data")
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / filename

if output_path.exists():
    output_path.unlink()

process = CrawlerProcess(settings={
    "USER_AGENT": "Mozilla/5.0",
    "LOG_LEVEL": logging.INFO,
    "FEEDS": {
        str(output_path): {"format": "json"},
    },
})

process.crawl(BookingSpider)
process.start()
```
